```python
import numpy as np

def sigmoid(z):
    return 1 / (1+np.exp(-z))
    
class Logisticregression:
    def __init__(self, learningrate=0.1, epochs=1000):
        self.learningrate = learningrate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.losses = []

    def fit(self , x ,y):
        n_samples , n_features = x.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        for i in range(self.epochs):
            p = np.dot(x,self.weights)+self.bias
            y_pred = sigmoid(p)
            loss = - (1/n_samples) * np.sum(y * np.log(y_pred + 10e-9) + (1-y) * np.log(1-y_pred + 10e-9))
            self.losses.append(loss)

            dW = (2/(n_samples)) * np.dot(x.T, (y_pred - y))
            dC = (2/(n_samples)) * np.sum((y_pred - y ))

            self.weights -= self.learningrate * dW
            self.bias -= self.learningrate * dC

        print(self.losses[-1])

    def predict(self , x):
        p = np.dot(x, self.weights) + self.bias
        y_pred = sigmoid(p)
        return [1 if i>0.5 else 0 for i in y_pred]


x = np.array([[1,2],[2,3],[2,3]])
y = np.array([1,0,0])

model = Logisticregression(learningrate = 0.1 , epochs =1000)
model.fit(x,y)
print(model.predict(x))
print(model.weights)
print(model.bias)
```
