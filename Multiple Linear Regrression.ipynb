```python
# multiple linear regression from scratch
import numpy as np

# Features
X=np.array([
    [1,10,22],
    [2,12,25],
    [3,13,28],
    [4,16,30],
    [5,18,35]
])

# Target
y=np.array([20000,25000,30000,40000,50000])

# step 2 : add Bias(Intercept)
ones = np.ones((X.shape[0],1))
X = np.hstack((ones,X))

# step 3 : Apply Normal Equation
XT = X.T                          # X transpose
XTX = np.dot(XT,X)                # X^T X
XTX_inv = np.linalg.inv(XTX)      # (X^T X)^-1
XTy= np.dot(XT,y)                # X^T y

beta = np.dot(XTX_inv, XTy)      # final Coefficient

# Step 4 : Output

print("Coefficient (Beta) : ")
print(beta)

# step 5: Prediction 
# example : Xxperience = 6 , Education =20, Age=40

new_data = np.array([1,6,20,40])

prediction = np.dot(new_data, beta)

print("Predicted Salary :", prediction )
```
